```python
MMBTU_TO_MJ = 1055.06 # 1mmBtu = 1055.06MJ
HHV1 = 55.6 # MJ/KgCH4 (El Abbadi, Preprint)
cf = 8322 # 95% Capacity Factor(cf) in hours
t = 20 # lifetime of membrane technology in years
wacc = 0.1 # WACC is the Weighted Average Cost of Capital (WACC). A 10% WACC is assumed.
P = 1 # Pressure(P) atm
T = 274.15 # Temperature(T) K
R = 0.0821 # Ideal gas constant(R) L.atm/mol.K
M = 16 # Molar mass(M) of CH4 g/mol
capex = 1_977_000 # ($)
om = 141_809 # O&M cost ($/yr)
vc = 55_837 # Variable cost ($/yr)
eff = 0.98 # Methane recovery effficiency (%)
mc = 0.55 # Methane content in biogas (%)
capacity = 127 # Nm3/hr
```

```python
"""
Capex $ ---> $/mmBtu
    Capex = $ 1,977,000
"""

def cal_rng_flow(mc,capacity,eff):
    """Calculates RNG flowrate in Nm3/h"""
    rng_flow = mc*capacity*eff
    return rng_flow
rng_flow = cal_rng_flow(mc,capacity,eff)
print("RNG flowrate(Nm3/h) = ", rng_flow)
```

```python
def cal_rng_flow_annual(rng_flow,cf):
    """Calculates RNG flowrate in Nm3/yr"""
    rng_flow_annual = rng_flow*cf
    return rng_flow_annual
rng_flow_annual = cal_rng_flow_annual(rng_flow,cf)
print("RNG flowrate(Nm3/yr) = ", rng_flow_annual)
```

```python
def cal_HHV(P,M,R,T,HHV1): 
    """Calculates HHV of methane in MJ/Nm3"""
    HHV2 = ((P*M)/(R*T))*HHV1
    return HHV2
HHV2 = cal_HHV(P,M,R,T,HHV1)
print("HHV(MJ/Nm3) = ", HHV2)
```

```python
def cal_rng_flow_annual_mmbtu(rng_flow_annual,HHV2,MMBTU_TO_MJ):
    """Calculates RNG flowrate in mmBtu/yr"""
    rng_flow_annual_mmbtu = (rng_flow_annual*HHV2)/MMBTU_TO_MJ
    return rng_flow_annual_mmbtu
rng_flow_annual_mmbtu = cal_rng_flow_annual_mmbtu(rng_flow_annual,HHV2,MMBTU_TO_MJ)
print("RNG flowrate(mmBtu/yr) = ", rng_flow_annual_mmbtu)
```

```python
def cal_crf(t,wacc):
    """Calculates CRF"""
    crf = wacc/(1-(1+wacc)**-t)
    return crf
crf = cal_crf(t,wacc)
print("CRF = ",crf)
```

```python
def cal_annual_capex(capex,crf):
    """Calculates annualized capex in $/yr"""
    annual_capex = capex*crf
    return annual_capex
annual_capex = cal_annual_capex(capex,crf)
print("Annualized Capex($/yr) = ",annual_capex)
```

```python
def cal_levelized_capex(annual_capex,rng_flow_annual_mmbtu):
    """Calculates levelized capex in $/mmBtu"""
    levelized_capex = annual_capex/rng_flow_annual_mmbtu
    return levelized_capex
levelized_capex = cal_levelized_capex(annual_capex,rng_flow_annual_mmbtu)
print("levelized Capex($/mmBtu) = ",levelized_capex)
```

```python
def cal_levelized_om(om,rng_flow_annual_mmbtu):
    """Calculates levelized O&M in $/mmBtu"""
    levelized_om = om/rng_flow_annual_mmbtu
    return levelized_om
levelized_om = cal_levelized_om(om,rng_flow_annual_mmbtu)
print("levelized O&M($/mmBtu) = ",levelized_om)
```

```python
def cal_levelized_vc(vc,rng_flow_annual_mmbtu):
    """Calculates levelized capex in $/mmBtu"""
    levelized_vc = vc/rng_flow_annual_mmbtu
    return levelized_vc
levelized_vc = cal_levelized_vc(vc,rng_flow_annual_mmbtu)
print("levelized Variable Cost($/mmBtu) = ",levelized_vc)
```

```python
Total_levelized_cost = levelized_capex+levelized_om+levelized_vc
print("Total levelized cost($/mmBtu) = ",Total_levelized_cost)
```

```python
"""
Electrigaz report already has energy cost incorporated in their variable cost.
For a scenario where there isn't annual electricity cost but just these parameters: 
"""
cap = 100 # Capacity in m3/h
ec = 0.24 # Energy consumption in kWh/m3
ep = 0.18 # Electricity price in $/kWh

"""Calculates annual energy cost"""
def cal_annual_energy(ec,cap,ep,cf):
    annual_energy = cap*ec*ep*cf
    return annual_energy
annual_energy =  cal_annual_energy(ec,cap,ep,cf)
print("Annual energy cost($/yr) = ",annual_energy)
```
